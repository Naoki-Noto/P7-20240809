```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem import Descriptors
```

```python
data = pd.read_csv(f'source/data_25286/zinc_data.csv')

duplicated_smiles = data[data.duplicated(subset='SMILES')]
print("duplication count:", len(duplicated_smiles))

data_unique = data.drop_duplicates(subset='SMILES')
data_unique
```

```python
def get_bertzct(smiles):
    molecule = Chem.MolFromSmiles(smiles)
    if molecule is not None:
        return Descriptors.BertzCT(molecule)
    else:
        return None

def get_kappa3(smiles):
    molecule = Chem.MolFromSmiles(smiles)
    if molecule is not None:
        return Descriptors.Kappa3(molecule)
    else:
        return None

data_unique['BertzCT'] = data_unique['SMILES'].apply(get_bertzct)
data_u = data_unique.dropna(subset=['BertzCT'])
data_u['Kappa3'] = data_u['SMILES'].apply(get_kappa3)
data_u2 = data_u.dropna(subset=['Kappa3'])
data_u2.to_csv("result/rdkit_25286.csv", index = False)
data_u2
```
